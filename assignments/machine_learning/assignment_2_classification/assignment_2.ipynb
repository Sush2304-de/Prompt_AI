```python
my_tuple = (10, 20, 30, 20, 40, 10, 50, 30)
repeated = []
for item in my_tuple:
    if my_tuple.count(item) > 1 and item not in repeated:
        repeated.append(item)
print("Repeated items:", tuple(repeated))
```

```python
students = (
    ("Amit", "CSE", (85, 78, 92, 88)),
    ("Riya", "CSE", (90, 88, 95, 91)),
    ("Rahul", "ECE", (70, 75, 68, 72)),
    ("Sneha", "CSE", (92, 91, 89, 94)),
    ("Arjun", "ECE", (82, 79, 85, 88))
)

print("a. Name and Department:")
for name, dept, marks in students:
    print(name, "-", dept)

print("\nb. & c. Total and Average:")
for name, dept, marks in students:
    total = sum(marks)
    avg = total / len(marks)
    print(name, "Total:", total, "Average:", avg)

highest = max(students, key=lambda x: sum(x[2]) / len(x[2]))
print("\nd. Highest Average:", highest[0])

lowest = min(students, key=lambda x: sum(x[2]) / len(x[2]))
print("e. Lowest Average:", lowest[0])

print("\nf. Average greater than 85:")
for name, dept, marks in students:
    avg = sum(marks) / len(marks)
    if avg > 85:
        print(name)

cse = [sum(marks) / len(marks) for name, dept, marks in students if dept == "CSE"]
print("\ng. Average of CSE students:", sum(cse) / len(cse))
```

```python
products = (
    ("Laptop", 55000, 12),
    ("Mobile", 25000, 25),
    ("Tablet", 30000, 18),
    ("Monitor", 15000, 10),
    ("Keyboard", 2000, 40)
)

ascending = sorted(products, key=lambda x: x[1])
print("a. Price Ascending:")
print(ascending)

descending = sorted(products, key=lambda x: x[1], reverse=True)
print("\nb. Price Descending:")
print(descending)

stock = sorted(products, key=lambda x: x[2])
print("\nc. Stock Quantity:")
print(stock)

expensive = max(products, key=lambda x: x[1])
print("\nd. Most Expensive:", expensive)

highest_stock = max(products, key=lambda x: x[2])
print("e. Highest Stock:", highest_stock)
```
