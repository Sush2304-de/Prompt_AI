```python
A = {10, 20, 30, 40, 50}
B = {30, 40, 50, 60, 70}

print("Union:", A | B)
print("Intersection:", A & B)
print("Difference A - B:", A - B)
print("Difference B - A:", B - A)
print("Symmetric Difference:", A ^ B)
print("Does 30 exist in A?", 30 in A)
```

```python
python_students = {"Amit", "Riya", "Rahul", "Sneha", "Ankit"}
java_students = {"Riya", "Rahul", "Priya", "Ankit", "Karan"}

print("Students enrolled in both courses:")
print(python_students & java_students)

print("Students enrolled only in Python:")
print(python_students - java_students)

print("Students enrolled only in Java:")
print(java_students - python_students)

print("Students enrolled in either Python or Java but not both:")
print(python_students ^ java_students)

print("Total number of unique students:")
print(len(python_students | java_students))
```

```python
A = {1, 2, 3, 4, 5, 6}
B = {2, 3, 4}
C = {7, 8}

print("Is B a subset of A?", B.issubset(A))
print("Is A a superset of B?", A.issuperset(B))
print("Is C a subset of A?", C.issubset(A))
print("Are A and C disjoint?", A.isdisjoint(C))
print("Are B and C disjoint?", B.isdisjoint(C))
```

```python
A = {10, 20, 30}

print("Original set:", A)

# add()
A.add(40)
print("After add(40):", A)

# update()
A.update([50, 60])
print("After update([50, 60]):", A)

# remove()
A.remove(20)
print("After remove(20):", A)

# discard()
A.discard(100)
print("After discard(100):", A)

# pop()
x = A.pop()
print("Element removed by pop():", x)
print("Set after pop():", A)

# clear()
A.clear()
print("After clear():", A)
```

```python
python = {"Amit", "Riya", "Rahul", "Sneha"}
java = {"Rahul", "Priya", "Karan", "Sneha"}
sql = {"Amit", "Rahul", "Priya", "Ankit"}

# 1. Employees who know all three technologies
print("Employees who know all three:")
print(python & java & sql)

# 2. Employees who know Python and Java
print("Employees who know Python and Java:")
print(python & java)

# 3. Employees who know Python but not Java
print("Employees who know Python but not Java:")
print(python - java)

# 4. Employees who know at least one technology
print("Employees who know at least one technology:")
print(python | java | sql)

# 5. Employees who know exactly two technologies
exactly_two = (
    (python & java - sql)
    | (python & sql - java)
    | (java & sql - python)
)

print("Employees who know exactly two technologies:")
print(exactly_two)
```
